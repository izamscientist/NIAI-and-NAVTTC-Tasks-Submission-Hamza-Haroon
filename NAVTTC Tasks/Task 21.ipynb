# Programiz: Python Lists Tutorial
All examples organized by section.

```python
# 1. Creating Lists
cart = ["T-shirt", "Lamp", "Pen"]
print(cart)

my_list = [1, "Python", 3.14]
print(my_list)

vowels_list = list("aeiou")
print(vowels_list)
```

```python
# 2. Indexing
languages = ["Python", "Swift", "C++"]
print('First:', languages[0])
print('Last:', languages[-1])
```

```python
# 3. Adding & Modifying
cart = ["T-shirt", "Lamp", "Pen"]
cart[1] = "Shoes"
cart.append("Book")
cart.insert(2, "Headphones")
print(cart)
```

```python
# 4. Removing Items
cart = ["T-shirt", "Lamp", "Pen", "Book"]
cart.remove("Pen")
popped = cart.pop()
print('After remove & pop:', cart)
print('Popped item:', popped)
```

```python
# 5. Shallow Copying
fav_items = ["T-shirt", "Lamp", "Pen"]
cart = fav_items.copy()
fav_items.append("Book")
print('fav_items:', fav_items)
print('cart:', cart)
```

```python
# 6. Membership & Iteration
cart = ["T-shirt", "Lamp", "Pen"]
print('Is Lamp in cart?:', 'Lamp' in cart)

for item in cart:
    print(item)
```
