```python
# 1. Creating Tuples, Packing & Unpacking
empty_tuple = ()
int_tuple = (1, 2, 3)
mixed_tuple = (1, "Hello", 3.4)
nested_tuple = ("mouse", [8, 4, 6], (1, 2, 3))

# Tuple packing (parentheses are optional)
packed_tuple = 3, 4.6, "dog"
print("Packed:", packed_tuple)

# Tuple unpacking
a, b, c = packed_tuple
print("Unpacked:", a, b, c)

# Creating a single-element tuple requires a trailing comma
not_a_tuple = ("hello")      # class 'str'
single_tuple = ("hello",)    # class 'tuple'
print("Single element tuple:", single_tuple, type(single_tuple))
```

```python
# 2. Accessing Elements (Indexing & Slicing)
my_tuple = ('p', 'e', 'r', 'm', 'i', 't')
print("First item [0]:", my_tuple[0])
print("Last item [-1]:", my_tuple[-1])
print("Second last [-2]:", my_tuple[-2])

# Nested indexing
n_tuple = ("mouse", [8, 4, 6], (1, 2, 3))
print("Nested n_tuple[0][3]:", n_tuple[0][3])  # 's'
print("Nested n_tuple[1][1]:", n_tuple[1][1])  # 4

# Slicing
prog_tuple = ('p', 'r', 'o', 'g', 'r', 'a', 'm', 'i', 'z')
print("Slice [1:4]:", prog_tuple[1:4])
print("Slice [:-7]:", prog_tuple[:-7])
print("Slice [7:]: ", prog_tuple[7:])
print("Slice [:]:  ", prog_tuple[:])
```

```python
# 3. Immutability, Modifications & Deletion
my_tuple = (4, 2, 3, [6, 5])

# Tuples are immutable, so item assignment fails:
try:
    my_tuple[1] = 9
except TypeError as e:
    print("Assignment error:", e)

# But mutable nested objects inside tuples CAN be modified:
my_tuple[3][0] = 9
print("Modified nested list:", my_tuple)

# Reassignment is valid (creates a new reference)
my_tuple = ('p', 'r', 'o', 'g', 'r', 'a', 'm', 'i', 'z')
print("Reassigned:", my_tuple)

# Concatenation (+) and Repetition (*)
print("Concatenation:", (1, 2, 3) + (4, 5, 6))
print("Repetition:", ("Repeat",) * 3)

# Deleting the whole tuple using 'del'
temp_tuple = (1, 2, 3)
del temp_tuple
```

```python
# 4. Tuple Methods (count & index)
fruit_tuple = ('a', 'p', 'p', 'l', 'e')
print("Count of 'p':", fruit_tuple.count('p'))
print("Index of 'l':", fruit_tuple.index('l'))
```

```python
# 5. Membership, Iteration & Built-in Functions
fruit_tuple = ('a', 'p', 'p', 'l', 'e')

# Membership test
print("'a' in fruit_tuple:", 'a' in fruit_tuple)
print("'b' in fruit_tuple:", 'b' in fruit_tuple)
print("'g' not in fruit_tuple:", 'g' not in fruit_tuple)

# Iteration
print("\nIterating:")
for name in ('John', 'Kate'):
    print("Hello", name)

# Built-ins with tuples
numbers = (4, 2, 8, 1, 6)
print("\nBuilt-in functions:")
print("len():", len(numbers))
print("max():", max(numbers))
print("min():", min(numbers))
print("sum():", sum(numbers))
print("sorted():", sorted(numbers))  # Returns a new sorted list
print("tuple('hello'):", tuple('hello'))
```
