```python
# 1. Creating Strings (Single, Double, and Multiline Quotes)
name = "Python"
print(name)

message = 'I love Python.'
print(message)

# Multiline string using triple quotes
multiline_str = """
Never gonna give you up
Never gonna let you down
"""
print(multiline_str)
```

```python
# 2. Accessing String Characters (Indexing and Slicing)
greet = 'hello'

# Positive indexing
print("greet[0]:", greet[0])
print("greet[1]:", greet[1])

# Negative indexing
print("greet[-1]:", greet[-1])
print("greet[-2]:", greet[-2])

# Slicing [start:end]
text = 'programiz'
print("Slice text[1:4]:", text[1:4])
print("Slice text[:3]: ", text[:3])
print("Slice text[3:]: ", text[3:])
print("Slice text[:]:  ", text[:])
```

```python
# 3. String Immutability and Reassignment
message = 'Hola Amigos'

# Strings are immutable; modifying in-place raises a TypeError:
try:
    message[0] = 'H'
except TypeError as e:
    print("Mutation error:", e)

# Reassigning to a new string reference works:
message = 'Hello Friends'
print("Reassigned string:", message)
```

```python
# 4. String Operations (Comparison, Concatenation, Membership & Length)
# Comparing strings
str1 = "Hello, world!"
str2 = "I love Python."
str3 = "Hello, world!"
print("str1 == str2:", str1 == str2)
print("str1 == str3:", str1 == str3)

# String concatenation (+)
greet = "Hello, "
name = "Jack"
result = greet + name
print("Concatenated:", result)

# Membership test (in, not in)
print("'a' in 'program':", 'a' in 'program')
print("'at' not in 'battle':", 'at' not in 'battle')

# Length of string
print("Length of 'Hello':", len('Hello'))

# Iterating through a string
print("\nIterating:")
for letter in 'Hello':
    print(letter)
```

```python
# 5. Common String Methods
text = 'Python Programming'

print("lower():", text.lower())
print("upper():", text.upper())
print("title():", text.title())
print("replace():", text.replace('Python', 'Java'))
print("find('Prog'):", text.find('Prog'))
print("startswith('Py'):", text.startswith('Py'))
print("split():", text.split(' '))

# Join
words = ['Python', 'is', 'fun']
print("join():", ' '.join(words))
```

```python
# 6. Escape Sequences and String Formatting (f-strings)
# Escape characters
quote = "He said, \"What's there?\""
print("Escaped quotes:", quote)

multiline_escaped = "Line 1\nLine 2\tIndented"
print(multiline_escaped)

# f-Strings (formatted string literals)
person = 'Cathy'
country = 'UK'
print(f"{person} is from {country}.")
```
