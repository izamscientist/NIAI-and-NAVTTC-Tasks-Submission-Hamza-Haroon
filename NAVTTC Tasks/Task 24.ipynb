```python
# 1. Creating a Set & Duplicate Handling
# Set of integer types
student_id = {112, 114, 116, 118, 115}
print("Student ID:", student_id)

# Set of string types
vowel_letters = {'a', 'e', 'i', 'o', 'u'}
print("Vowel Letters:", vowel_letters)

# Set of mixed data types
mixed_set = {'Hello', 101, -2, 'Bye'}
print("Mixed Set:", mixed_set)

# Sets automatically eliminate duplicates
numbers = {2, 4, 6, 6, 2, 8}
print("Duplicates eliminated:", numbers)

# Creating an empty set (must use set(), {} creates an empty dict)
empty_set = set()
empty_dictionary = {}
print("type of empty_set:", type(empty_set))
print("type of empty_dictionary:", type(empty_dictionary))
```

```python
# 2. Adding and Updating Set Items
numbers = {21, 34, 54, 12}
print("Initial set:", numbers)

# add() adds a single element
numbers.add(32)
print("After add(32):", numbers)

# update() adds elements from another iterable (list, tuple, set)
companies = {'Lacoste', 'Ralph Lauren'}
tech_companies = ['apple', 'google', 'apple']
companies.update(tech_companies)
print("After update():", companies)
```

```python
# 3. Removing Items from a Set (discard, remove, pop, clear)
languages = {'Swift', 'Java', 'Python'}
print("Initial set:", languages)

# discard() removes an item; does NOT raise an error if item is absent
languages.discard('Java')
print("After discard('Java'):", languages)

# remove() removes an item; raises KeyError if item is absent
languages.remove('Python')
print("After remove('Python'):", languages)

# pop() removes and returns an arbitrary item
colors = {'red', 'yellow', 'green'}
popped_item = colors.pop()
print("Popped item:", popped_item)
print("Colors after pop():", colors)

# clear() empties the set
colors.clear()
print("After clear():", colors)
```

```python
# 4. Set Operations: Union, Intersection, Difference & Symmetric Difference
A = {1, 3, 5}
B = {1, 2, 3}

# Set Union (| operator or union() method)
print("Union using |:", A | B)
print("Union using union():", A.union(B))

# Set Intersection (& operator or intersection() method)
print("\nIntersection using &:", A & B)
print("Intersection using intersection():", A.intersection(B))

# Set Difference (- operator or difference() method)
print("\nDifference (A - B):", A - B)
print("Difference using difference():", A.difference(B))

# Set Symmetric Difference (^ operator or symmetric_difference() method)
print("\nSymmetric Difference using ^:", A ^ B)
print("Symmetric Difference using method:", A.symmetric_difference(B))
```

```python
# 5. Membership, Iteration & Built-in Functions
# Check if an element exists using 'in'
animals = {'cat', 'dog', 'tiger'}
print("'cat' in animals:", 'cat' in animals)
print("'elephant' in animals:", 'elephant' in animals)

# Iterating through a set
print("\nIterating through letters:")
letters = {'p', 'y', 't', 'h', 'o', 'n'}
for letter in letters:
    print(letter)

# Built-in functions with sets
numbers = {10, 20, 30, 40}
print("\nlen():", len(numbers))
print("max():", max(numbers))
print("min():", min(numbers))
print("sum():", sum(numbers))
print("sorted():", sorted(numbers))
```
