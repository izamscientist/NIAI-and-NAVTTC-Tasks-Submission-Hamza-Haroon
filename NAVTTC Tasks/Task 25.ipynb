```python
# 1. Creating a Dictionary
# Dictionary with keys and values of string type
capital_city = {"Nepal": "Kathmandu", "Italy": "Rome", "England": "London"}
print("Capital cities:", capital_city)

# Dictionary with keys of integer type
numbers = {1: "One", 2: "Two", 3: "Three"}
print("Numbers:", numbers)

# Dictionary with mixed keys and values
mixed_dict = {"name": "Bill", 1: [2, 4, 3]}
print("Mixed dictionary:", mixed_dict)

# Creating an empty dictionary
empty_dict = {}
print("Empty dictionary:", empty_dict, type(empty_dict))
```

```python
# 2. Accessing Elements (Bracket Notation vs. get())
student_info = {"name": "Harry", "age": 22, "course": "CS"}

# Accessing using square bracket []
print("Name:", student_info["name"])
print("Age:", student_info["age"])

# Accessing using get()
print("Course:", student_info.get("course"))

# Difference between [] and get() when a key does not exist:
print("get() with missing key:", student_info.get("grade"))  # Returns None
print("get() with default fallback:", student_info.get("grade", "Not Enrolled"))

try:
    print(student_info["grade"])
except KeyError as e:
    print("Square brackets [] raise KeyError when missing:", e)
```

```python
# 3. Adding and Modifying Dictionary Items
capital_city = {"Nepal": "Kathmandu", "Italy": "Rome"}
print("Initial dictionary:", capital_city)

# Adding a new key-value pair
capital_city["Japan"] = "Tokyo"
print("After adding Japan:", capital_city)

# Updating an existing key's value
capital_city["Italy"] = "Milan"
print("After updating Italy:", capital_city)
```

```python
# 4. Removing Items from a Dictionary (pop, popitem, del, clear)
student_info = {"name": "Harry", "age": 22, "course": "CS", "city": "London"}
print("Initial:", student_info)

# pop() removes the item with the specified key and returns its value
removed_course = student_info.pop("course")
print("Popped course:", removed_course)
print("After pop('course'):", student_info)

# popitem() removes and returns the last inserted key-value pair
last_item = student_info.popitem()
print("popitem() returned:", last_item)
print("After popitem():", student_info)

# del keyword deletes an item by key
del student_info["age"]
print("After del student_info['age']:", student_info)

# clear() removes all elements
student_info.clear()
print("After clear():", student_info)
```

```python
# 5. Common Dictionary Methods (keys, values, items, copy, update)
marks = {"Physics": 67, "Maths": 87, "English": 75}

# keys() returns a view of all keys
print("Keys:", marks.keys())

# values() returns a view of all values
print("Values:", marks.values())

# items() returns a view of (key, value) tuple pairs
print("Items:", marks.items())

# update() inserts specified items/dictionaries
marks.update({"Chemistry": 91, "Maths": 95})
print("After update():", marks)

# copy() returns a shallow copy
marks_copy = marks.copy()
marks_copy["Biology"] = 80
print("Original marks:", marks)
print("Copied marks:", marks_copy)
```

```python
# 6. Membership, Iteration & Built-ins
squares = {1: 1, 3: 9, 5: 25, 7: 49, 9: 81}

# Membership test (checks KEYS, not values)
print("1 in squares:", 1 in squares)
print("2 in squares:", 2 in squares)
print("49 in squares (value check with 'in' fails):", 49 in squares)
print("49 in squares.values():", 49 in squares.values())

# Iterating over keys
print("\nIterating through keys:")
for k in squares:
    print(f"Key: {k}, Value: {squares[k]}")

# Iterating over key-value pairs using .items()
print("\nIterating with .items():")
for key, value in squares.items():
    print(f"{key} -> {value}")

# Built-ins
print("\nlen():", len(squares))
print("sorted() keys:", sorted(squares))
```
