```python
# 1. Basic try...except Block
try:
    numerator = 10
    denominator = 0

    result = numerator / denominator
    print(result)
except ZeroDivisionError:
    print("Error: Denominator cannot be 0.")
```

```python
# 2. Catching Multiple / Specific Exceptions
try:
    even_numbers = [2, 4, 6, 8]
    print("Attempting to access index 5:")
    print(even_numbers[5])
except ZeroDivisionError:
    print("Denominator cannot be 0.")
except IndexError:
    print("Index Out of Bound.")

# Catching multiple exception types in a single tuple
try:
    value = int("not_a_number")
except (ValueError, TypeError) as e:
    print(f"Caught specific error: {e}")
```

```python
# 3. try...except...else Block
# The else block executes only if no exception is raised inside try
try:
    num = 4
    assert num % 2 == 0
except AssertionError:
    print("Not an even number!")
else:
    reciprocal = 1 / num
    print(f"Reciprocal of {num} is {reciprocal}")
```

```python
# 4. try...finally Block
# The finally block ALWAYS executes, regardless of whether an exception occurs
try:
    numerator = 10
    denominator = 0
    result = numerator / denominator
    print(result)
except ZeroDivisionError:
    print("Error: Denominator cannot be 0.")
finally:
    print("This is the finally block - cleanup code always runs here.")
```

```python
# 5. Raising Exceptions with the raise Keyword
def check_salary(salary):
    if salary < 0:
        raise ValueError("Salary cannot be negative.")
    return f"Valid salary: ${salary}"

try:
    print(check_salary(5000))
    print(check_salary(-1500))
except ValueError as e:
    print("Caught manual exception:", e)
```
