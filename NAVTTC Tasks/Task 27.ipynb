```python
# 1. Defining and Raising a Basic Custom Exception
# In Python, custom exceptions inherit from the built-in Exception class
class InvalidAgeException(Exception):
    """Raised when the input age is less than 18"""
    pass

def check_voting_eligibility(age):
    if age < 18:
        raise InvalidAgeException
    else:
        print(f"Age {age}: Eligible to Vote")

# Testing the custom exception
for test_age in [20, 16]:
    try:
        print(f"\nChecking age {test_age}...")
        check_voting_eligibility(test_age)
    except InvalidAgeException:
        print("Exception occurred: Invalid Age (must be 18 or older)")
```

```python
# 2. Customizing Exception Classes (Adding Attributes & Custom Messages)
# Overriding __init__ to store contextual details about what went wrong
class SalaryNotInRangeError(Exception):
    """Exception raised for errors in the input salary.

    Attributes:
        salary  -- input salary that caused the error
        message -- explanation of the error
    """
    def __init__(self, salary, message="Salary is not in the valid range (5000 to 15000)"):
        self.salary = salary
        self.message = message
        super().__init__(self.message)

def validate_salary(amount):
    if not (5000 <= amount <= 15000):
        raise SalaryNotInRangeError(amount)
    print(f"Salary accepted: ${amount}")

# Testing valid and invalid ranges
test_salaries = [8500, 3000]
for sal in test_salaries:
    try:
        validate_salary(sal)
    except SalaryNotInRangeError as err:
        print(f"Caught SalaryNotInRangeError: {err.salary} -> {err.message}")
```

```python
# 3. Customizing Exception String Output with __str__
class ValueTooSmallError(Exception):
    def __init__(self, value):
        self.value = value

    def __str__(self):
        return f"[Error] Entered value '{self.value}' is too small! Must be at least 10."

try:
    n = 4
    if n < 10:
        raise ValueTooSmallError(n)
except ValueTooSmallError as e:
    print(e)
```
