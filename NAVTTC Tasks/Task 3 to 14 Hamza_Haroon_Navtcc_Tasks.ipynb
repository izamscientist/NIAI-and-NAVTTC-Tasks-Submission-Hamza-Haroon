```python
# Task 5: Calculate the square root
num = 8
# Or take input: num = float(input('Enter a number: '))

num_sqrt = num ** 0.5
print('The square root of %0.3f is %0.3f' % (num, num_sqrt))
```

```python
# Task 6: Square root using cmath
import cmath

num = 1 + 2j
# Or take input: num = complex(input('Enter a complex number (e.g. 1+2j): '))

num_sqrt = cmath.sqrt(num)
print('The square root of {0} is {1:0.3f}+{2:0.3f}j'.format(num, num_sqrt.real, num_sqrt.imag))
```

```python
# Task 7: Convert Celsius to Fahrenheit
celsius = 37.5
fahrenheit = (celsius * 1.8) + 32

print('%0.1f degree Celsius is equal to %0.1f degree Fahrenheit' % (celsius, fahrenheit))
```

```python
# Task 8: Find the area of a triangle
a = 5
b = 6
c = 7

# Calculate semi-perimeter
s = (a + b + c) / 2

# Calculate area
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
print('The area of the triangle is %0.2f' % area)
```

```python
# Task 9: Quadratic equation solver
import cmath

a = 1
b = 5
c = 6

# Calculate discriminant
d = (b**2) - (4 * a * c)

# Find solutions
sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

print('The solutions are {0} and {1}'.format(sol1, sol2))
```

```python
# Task 10: Kilometers to miles
kilometers = float(input("Enter value in kilometers: "))
conv_fac = 0.621371

miles = kilometers * conv_fac
print('%0.2f kilometers is equal to %0.2f miles' % (kilometers, miles))
```

```python
# Task 11: Simple if condition
i = 10

if i > 15:
    print("10 is less than 15")

print("I am Not in if")
```

```python
# Task 12: if-else statement
i = 20

if i < 15:
    print("i is smaller than 15")
    print("i'm in if Block")
else:
    print("i is greater than 15")
    print("i'm in else Block")

print("i'm not in if and not in else Block")
```

```python
# Task 13: Nested if blocks
i = 10

if i == 10:
    # First nested check
    if i < 15:
        print("i is smaller than 15")
    # Second nested check
    if i < 12:
        print("i is smaller than 12 too")
    else:
        print("i is greater than 15")
```

```python
# Task 14: Multi-way branching
i = 20

if i == 10:
    print("i is 10")
elif i == 15:
    print("i is 15")
elif i == 20:
    print("i is 20")
else:
    print("i is not present")
```
