```python
# statistics
"""
Descriptive Statistics: Mean, Median, Mode, and Dispersion
Ready to run in a Colab notebook cell or as a standalone script.
"""

import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. Dataset Initialization
# ---------------------------------------------------------
# Sample speed data (from the W3Schools machine learning dataset)
speed = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])

print("=" * 45)
print("             DATASET SUMMARY")
print("=" * 45)
print(f"Data points: {speed}")
print(f"Total count (n): {len(speed)}")
print(f"Sorted data    : {np.sort(speed)}\n")

# ---------------------------------------------------------
# 2. Measures of Central Tendency
# ---------------------------------------------------------
# Mean: Sum of values divided by count
mean_val = np.mean(speed)

# Median: Middle value of the sorted sequence
median_val = np.median(speed)

# Mode: Most frequently occurring value
# scipy.stats.mode returns ModeResult(mode, count)
mode_result = stats.mode(speed, keepdims=False)
mode_val = mode_result.mode
mode_count = mode_result.count

print("=" * 45)
print("        CENTRAL TENDENCY MEASURES")
print("=" * 45)
print(f"Mean   : {mean_val:.2f}")
print(f"Median : {median_val:.2f}")
print(f"Mode   : {mode_val} (appeared {mode_count} times)\n")

# ---------------------------------------------------------
# 3. Measures of Dispersion & Position
# ---------------------------------------------------------
std_dev = np.std(speed)
variance = np.var(speed)
p25 = np.percentile(speed, 25)
p75 = np.percentile(speed, 75)
iqr = p75 - p25

print("=" * 45)
print("          DISPERSION & PERCENTILES")
print("=" * 45)
print(f"Variance (σ²)          : {variance:.2f}")
print(f"Standard Deviation (σ) : {std_dev:.2f}")
print(f"25th Percentile (Q1)   : {p25:.2f}")
print(f"75th Percentile (Q3)   : {p75:.2f}")
print(f"Interquartile Range    : {iqr:.2f}\n")

# ---------------------------------------------------------
# 4. Visualization in Google Colab
# ---------------------------------------------------------
plt.figure(figsize=(9, 4.5))
plt.hist(speed, bins=8, color="#4A90E2", edgecolor="black", alpha=0.7, label="Frequency")

# Overlay vertical lines for Mean, Median, and Mode
plt.axvline(mean_val, color="#E74C3C", linestyle="--", linewidth=2, label=f"Mean: {mean_val:.2f}")
plt.axvline(median_val, color="#2ECC71", linestyle="-.", linewidth=2, label=f"Median: {median_val:.2f}")
plt.axvline(mode_val, color="#F39C12", linestyle=":", linewidth=2.5, label=f"Mode: {mode_val}")

plt.title("Distribution with Mean, Median, and Mode", fontsize=13, pad=12)
plt.xlabel("Speed Values", fontsize=11)
plt.ylabel("Frequency", fontsize=11)
plt.legend(frameon=True)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
